## 1. Az adat beolvasása

A magyar havi infláció (HICP, éves változás, %) 2023 januárjától, az Eurostat adataiból. A fájl a notebook melletti `data/` mappában van.

```python
from pathlib import Path

import pandas as pd

utvonal = "data/hicp_osszesen.csv"
# Colabban nincs meg a data/ mappa, ott a kurzusoldalról töltjük le ugyanezt a fájlt.
if not Path(utvonal).exists():
    utvonal = "https://tubolyroli.github.io/ai-kurzus/notebookok/data/hicp_osszesen.csv"

adat = pd.read_csv(utvonal, sep=";")
adat.tail()
```

A cella alatt megjelenik a táblázat utolsó öt sora: a hónap és az éves inflációs ráta.

```python
csucs = adat.loc[adat["hicp_ev_ev"].idxmax()]
print(f"A legmagasabb érték: {csucs['hicp_ev_ev']}% ({csucs['honap']})")
print(f"Hónapok száma az adatban: {len(adat)}")
```

## 2. Ábra

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(9, 4))
ax.plot(pd.to_datetime(adat["honap"]), adat["hicp_ev_ev"], color="#252A51", linewidth=2)
ax.set_ylabel("Éves változás (%)")
ax.set_title("Magyar infláció (HICP), 2023–2025")
ax.grid(alpha=0.3)
plt.show()
```
